```python
import numpy as np
import pandas as pd
```

Pandas Series

It is a one dimensional labeled array capable of holding any data type. The axis labels are collectively called the index. It is vertical unlike arrays in numpy.

```python
labels = ['a', 'b', 'c']
myList = [10, 20, 30]
arr = np.array([10, 20, 30])
d = {1:10, 2:20, 3:30}
```

```python
pd.Series(myList)
```

0, 1, 2 are labels and 10, 20, 30 are data points.

```python
pd.Series(myList, index=labels)
```

These are custom labels.

```python
pd.Series(arr)
```

```python
pd.Series(d)
```

In case of dictonary the keys are automatically converted to labels and values to datapoints.
