```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r'anime.csv')
```

1. Make a new column for episode count

```python
df.loc[1]['Title']
```

```python
def episodeExtraction(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True: 
            data = data + i 
        if i == '(':
            check = True 
```

```python
df['Episode Count'] = df['Title'].apply(episodeExtraction)
```

```python
df['Episode Count'] = df['Episode Count'].str.replace(" eps", "")
```

```python
df['Episode Count'] = df['Episode Count'].astype(int)
```

```python
df.head()
```

2. Make a new column for time stamp

```python
df.loc[0]['Title']
```

```python
def timeExtraction(title):
    timeStamp = "";
    for i in range(len(title)):
        if title[i] == ")":
            for j in range(i + 1, i + 20):
                timeStamp += title[j]
            return timeStamp
```

```python
df['Total Time'] = df['Title'].apply(timeExtraction)
```

```python
df.head()
```

3. Which anime has the highest score

```python
df[df['Score'] == df['Score'].max()]['Title']
```

```python
from dateutil.relativedelta import relativedelta
from datetime import datetime

def calcTotalMonths(period):
    try:
        start_str, end_str = period.split(' - ')
        start_date = datetime.strptime(start_str, '%b %Y')
        end_date = datetime.strptime(end_str, '%b %Y')
        r = relativedelta(end_date, start_date)
        return r.years * 12 + r.months + 1
    except:
        return None

df['Months'] = df['Total Time'].apply(calcTotalMonths)
```
