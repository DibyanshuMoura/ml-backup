```python
import numpy as np
import pandas as pd
```

## Handling missing data

Identifying missing data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
```

```python
dataF = pd.DataFrame(data)
dataF
```

```python
dataF.isna()
```

```python
dataF.isna().sum()
```

```python
dataF.isna().any()
```

Removing missing data

```python
dataF
```

```python
dataF.dropna() #works in rows
```

```python
dataF.dropna(thresh=3) #The thresh value means minimum number of non 
# null values should be equal to the thresh value
# works in rows
```

Filling the missing data

```python
dataF
```

```python
dataF.fillna(0)
```

```python
values = {'A': 0, 'B': 100, 'C': 200, 'D': 300}
dataF.fillna(value=values) # fills column wise
```

```python
dataF.fillna(dataF.mean())
```
