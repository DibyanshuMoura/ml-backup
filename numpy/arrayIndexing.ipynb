```python
import numpy as np
```

Numpy indexing and slicing of vectors

```python
arr = np.arange(11, 21)
arr
```

```python
arr[6] #Indexing
```

```python
arr[1:5] #Slicing
```

```python
arr[:5]
```

```python
arr[:]
```

```python
arr[3:]
```

```python
arr_slice = arr[3::2] #3rd colon is for gap
arr_slice #Slicing returns the sliced array
```

Numpy indexing and slicing of matrix

```python
arr = np.arange(1, 31).reshape(6, 5)
arr
```

```python
arr[0]
```

```python
arr[0][0] #Indexing
```

```python
arr[0,0] #Indexing
```

```python
slic = arr[:2, 1:3] #Combined row-col slicing
slic
```

```python
slic = arr[3:, 3:]
slic
```

Boolean Indexing

```python
arr = np.arange(11, 21)
arr
```

```python
bool_idx = arr % 2 == 0
bool_idx
```

```python
arr = arr[bool_idx]
arr
```
