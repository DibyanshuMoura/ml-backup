You cannot stack 2 matrices cause they are already
stacked, you can only stack vectors

```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])
```

```python
np.vstack((a, b)) # Vertical Stack, row-wise
```

```python
np.hstack((a, b)) # Horizonatl Stack, col-wise
```

```python
np.column_stack((a, b)) # Column Stack
```
