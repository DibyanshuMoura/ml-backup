27. Create an array [10, 20, 30, 40, 50]. Make a copy of it, change the first element of the copy to 1, and check whether the original changed.

```python
a = np.array([10, 20, 30, 40, 50])
b = a
```

```python
b[0] = 1
```

```python
a
```

```python
b
```

This happened because directly copying creates a shallow copy which means b is pointing to the same memory location to which a is pointing.

```python
a = np.array([10, 20, 30, 40, 50])
b = a.copy()
```

```python
b[0] = 1
```

```python
a
```

```python
b
```

This is called a deep copy another array is created in memory to which b is pointing not the same as a.
