import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt

Line chart

```python
x = [1, 2, 3, 4]
y = [10, 15, 10, 10]

plt.plot(x, y)
plt.figure(figsize=(4, 4))
plt.show()
```

```python
x = ['Mon', 'Tues', 'Wed', 'Thurs', 'Fri']
y = [10, 15, 7, 20, 12]

plt.plot(x, y)
plt.title("Bakery Sales this week")
plt.xlabel("Days")
plt.ylabel("Number of sales")
plt.grid()

plt.show()
```

```python
months = [1, 2, 3, 4]
sales = [1200, 1000, 2000, 1800]
plt.plot(months, sales, color='red', linestyle='--', linewidth=2, marker='o', label='Sales of 4 months')
plt.xlabel("Month of the year")
plt.ylabel("Sales")
plt.title("Sales of 4 months")
plt.grid(color='blue', linestyle=':', linewidth=1)
plt.legend(loc='lower right', fontsize=10)
plt.xlim(1, 4)
plt.ylim(0, 2200)
plt.xticks([1, 2, 3, 4], ['M1', 'M2', 'M3', 'M4'])
plt.show()
```

```python
x = [1, 2, 3, 4]
y1 = [10, 15, 10, 10]
y2 = [20, 25, 20, 20]

plt.plot(x, y1, label='Sales Jan')
plt.plot(x, y2, label='Sales Feb')

plt.title("Sales of 2 months")
plt.xlabel("Days Quaterly")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

Bar Chart

```python
x = ['A', 'B', 'C', 'D', 'E']
y = [40, 20, 35, 30, 50]

plt.bar(x, y)
plt.title("Analysis")

plt.show()
```

Histogram

```python
import random
data = [random.randint(1, 10) for _ in range(100)]

plt.hist(data, bins=5)
plt.title("Histotical")

plt.show()
```

Pie Chart

```python
categories = ['A', 'B', 'C', 'D', 'E']
sales = [40, 20, 35, 30, 50]

plt.pie(sales, labels=categories, autopct='%1.1f%%', startangle=90)

plt.show()
```

Scatter Plot

Used to find relationship bw variables

```python
x = [10, 20, 30, 34, 23]
y = [23, 32, 21, 43, 11]

plt.scatter(x, y)
plt.title("Scatter Plot")

plt.show()
```

Subplots

Used to show multiple charts in one figure

```python
categories = ['Mon', 'Tues', 'Wed', 'Thur', 'Fri']
sales = [40, 20, 35, 30, 50]

y1 = [10, 15, 10, 10]
y2 = [20, 25, 20, 12]

plt.figure(figsize=(8, 5))

# 1st sub plot
plt.subplot(1, 2, 1) # row, column, position
plt.bar(categories, sales)
plt.title("Daily Sales")
plt.xlabel("Week Days")
plt.ylabel("Sales")

# 2nd sub plot
plt.subplot(1, 2, 2)
plt.scatter(y1, y2)
plt.title("Users Example")
plt.xlabel("User 1")
plt.ylabel("User 2")

plt.show()
```
