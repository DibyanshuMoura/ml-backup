```python
def evaluate(val):
    if val < 2:
        return 'Low'
    elif 2 <= val < 4:
        return 'Med'
    else:
        return 'High'
df['study'] = df['study_hours'].apply(evaluate)
```

```python
df.head()
```

```python
def evaluate(val):
    if val < 60:
        return 'Poor'
    elif 60 <= val < 80:
        return 'Average'
    else:
        return 'Excellent'
df['performance'] = df['final_score'].apply(evaluate)
```

```python
df.head()
```

```python
def evaluate(val):
    if val < 75:
        return 'Low'
    elif 75 <= val < 90:
        return 'Moderate'
    else:
        return 'High'
df['attendanceC'] = df['attendance'].apply(evaluate)
```
