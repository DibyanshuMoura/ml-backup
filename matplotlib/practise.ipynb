```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Months': ['Jan', 'Feb', 'Mar', 'Apr'],
    'Sales': [12000, 11000, 13000, 25000]
}
df = pd.DataFrame(data)
df
```

```python
plt.bar(df['Months'], df['Sales'])
plt.title("Pandas")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.savefig("pratise.png") # to save the image of the chart in device locally

plt.show()
```
